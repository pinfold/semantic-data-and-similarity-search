## Semantic Data — Explanation and Python Demonstration
### What is Semantic Data?

Semantic data is data that is structured in a way that explicitly encodes meaning and relationships, not just values in rows and columns.

Instead of:

| Person | Employer | City |
| -------- | -------- | -------- |
| Alice | Acme | Portland |

Semantic data represents this as machine-understandable facts, typically using:

* URIs (global identifiers)
* Triples: (subject, predicate, object)
* Ontologies (shared vocabularies)
* Graphs instead of tables

Example [semantic triple](https://en.wikipedia.org/wiki/Semantic_triple):

<http://example.org/Alice> <http://schema.org/worksFor> <http://example.org/AcmeCorp>

#### The Anatomy of a Triple
In the world of Linked Data, every triple consists of three parts: 

**Subject:** http://example.org/Alice

The "thing" the statement is about.

**Predicate (Property):** http://schema.org/worksFor

The specific relationship or characteristic being described.

**Object:** http://example.org/AcmeCorp

The value or the other "thing" linked to the subject. 
#### Why use URLs instead of names?
Instead of just saying "Alice," the triple uses a URI (Uniform Resource Identifier). This is crucial for two reasons: 

**Uniqueness:** There are millions of people named Alice. Using a specific URL (like example.org/Alice) ensures the computer knows exactly which Alice we are talking about.

Interoperability: By using schema.org/worksFor, the data follows a globally recognized vocabulary. Any system in the world that understands Schema.org now knows exactly what the relationship between Alice and Acme Corp is. 
#### The "Graph" Concept
If you add another triple, such as Alice livesIn Portland, you start to build a Knowledge Graph. Because "Alice" is the subject in both statements, a computer can automatically link those two facts together to realize that an employee of Acme Corp lives in Portland, even if that specific fact was never explicitly written down.

This allows machines to:

✅ Understand relationships
✅ Merge data from multiple sources
✅ Infer new facts
✅ Query across domains

This is the foundation of:

* Knowledge Graphs
* Linked Data
* Semantic Web
* RDF / OWL / SPARQL
* Enterprise data integration
* AI + reasoning systems

Key Concepts

| Concept   | Meaning                        |
| --------- | ------------------------------ |
| URI       | Global identifier for things   |
| Triple    | (subject, predicate, object)   |
| RDF       | Resource Description Framework |
| Ontology  | Formal vocabulary + meaning    |
| SPARQL    | Query language for RDF graphs  |
| Inference | Deriving new facts             |


## Python Demo: Working with Semantic Data (RDF)

We’ll use rdflib, the standard Python library for semantic data.

**Resource Description Framework (RDF)** is a [World Wide Web Consortium](https://www.w3.org/RDF/) (W3C) standard for representing interconnected data on the web, using simple subject-predicate-object "triples" to describe resources, enabling data to be easily shared, merged, and understood by machines, crucial for knowledge graphs and data integration.

* **Graph:** The core container that holds RDF triples (Subject $\rightarrow$ Predicate $\rightarrow$ Object).
* **Namespace:** A tool to create URIs (Uniform Resource Identifiers) easily by appending a name to a base
* **URL.URIRef & Literal:** The building blocks of data. URIRef represents a web resource (a URL), while Literal represents raw data like strings, dates, or numbers.
* **Predefined Namespaces:** The code imports common vocabularies like RDF (core logic), FOAF (Friend of a Friend, for social data), and XSD (XML Schema types for data like integers or dates).

```python
from rdflib import Graph, Namespace, URIRef, Literal
from rdflib.namespace import RDF, FOAF, XSD

# Create a graph
g = Graph()

# Define namespaces
EX = Namespace("http://example.org/")
SCHEMA = Namespace("http://schema.org/")

# Bind prefixes (for readability)
g.bind("ex", EX)
g.bind("schema", SCHEMA)
g.bind("foaf", FOAF)
```

## Step 2 — Add Semantic Facts (Triples)

```python
# Define people and organizations as URIs
alice = EX.Alice
bob = EX.Bob
acme = EX.AcmeCorp
portland = EX.Portland

# Add facts
g.add((alice, RDF.type, FOAF.Person))
g.add((bob, RDF.type, FOAF.Person))
g.add((acme, RDF.type, SCHEMA.Organization))
g.add((portland, RDF.type, SCHEMA.City))

# Add properties
g.add((alice, FOAF.name, Literal("Alice")))
g.add((bob, FOAF.name, Literal("Bob")))

g.add((alice, SCHEMA.worksFor, acme))
g.add((bob, SCHEMA.worksFor, acme))

g.add((alice, SCHEMA.homeLocation, portland))
g.add((bob, SCHEMA.homeLocation, portland))
```

You have built a Knowledge Graph. Instead of a flat table (like Excel), you have created a web of relationships. 

By running the code, you have logically linked your entities:

* Alice and Bob are types of *FOAF.Person*.
* They both work for *AcmeCorp*.
* They both live in *Portland*.

Because they share a workplace and a city, your graph now contains "hidden" knowledge: Alice and Bob are colleagues who live in the same location.

## Step 3 — View the Graph (Turtle Format)

```python
print(g.serialize(format="turtle"))
```

This shows your knowledge graph in a human-readable semantic format.

## Step 4 — Query with SPARQL
Find everyone who works for AcmeCorp

```python
query = """
PREFIX schema: <http://schema.org/>
PREFIX foaf: <http://xmlns.com/foaf/0.1/>
PREFIX ex: <http://example.org/>

SELECT ?personName
WHERE {
  ?person schema:worksFor ex:AcmeCorp .
  ?person foaf:name ?personName .
}
"""

for row in g.query(query):
    print(row.personName)
```

## Step 5 — Semantic Reasoning Example

Add a rule-like structure manually:

If someone works for AcmeCorp, then they are an Employee.

```python
EMPLOYEE = EX.Employee

# Add inferred facts manually (simulating reasoning)
for person in [alice, bob]:
    g.add((person, RDF.type, EMPLOYEE))
```

Now query for Employees:

```python
query = """
PREFIX ex: <http://example.org/>

SELECT ?person
WHERE {
  ?person a ex:Employee .
}
"""

for row in g.query(query):
    print(row.person)
```
